```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
spotify_df = pd.read_csv("../data//archive/Cleaned_Spotify_2024_Global_Streaming_Data.csv")
spotify_df.head()
```

```python
yearly_avg = spotify_df.groupby("Release Year")["Total Streams (Millions)"].mean().round(2)
yearly_avg
```

```python
plt.figure(figsize=(12,6))
plt.plot(
    yearly_avg.index, 
    yearly_avg.values,
    marker = "o", 
    linestyle = "--", 
    color = "#2C44AD",
    linewidth = 0.5
    )


plt.xlabel("Release Year")
plt.ylabel("Average Total Streams")
plt.title("Did Listeners Stream Less in 2022?", fontsize = 18, pad= 20)

ax = plt.gca()
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

ax.axvline(
    2022, 
    color= "#934848",
    linewidth = 0.5 ,
    alpha = 0.7 
    )

ax.text(
    2022+0.05, 
    2225,
    "Dip In Total Streams",
    color = "#934848", 
    fontsize=8
)

plt.ylim(0, 3500)

plt.tight_layout()

plt.show()
```
