```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
spotify_df = pd.read_csv("../data//archive/Cleaned_Spotify_2024_Global_Streaming_Data.csv")
spotify_df.head()
```

```python
category = "Platform Type"
values = "Total Streams (Millions)"

platform_totals = spotify_df.groupby(category)[values].sum()
platform_totals
```

```python
plt.figure(figsize=(12,8))
plt.pie(
    platform_totals,
    labels=None,
    startangle=110,
    colors=["gray", "#7fcdbb",],
    autopct= "%1.1f%%"
    )

plt.title("Near Even Split Between Premium and Free User")

plt.legend(
    labels= platform_totals.index,
    loc = "lower right",
    title= "Platform"
)

plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(figsize =(8,8))



plt.figure(figsize=(12,8))
wedges, texts, autotexts = ax.pie(
    platform_totals,
    autopct= "%1.1f%%", 
    startangle=110,
    colors=["gray", "#7fcdbb",],
    )

ax.set_title(
    "Near Even Split Between Premium and Free Users", 
    fontsize=35,
    pad=15,
    loc="center"
    )

for autotext in autotexts:
    autotext.set_fontsize(25)
    autotext.set_color("white")
    autotext.set_weight("bold")

ax.legend(
    wedges, 
    platform_totals.index,
    loc="lower center",
    bbox_to_anchor = (0.5, 0.95),
    ncol = 2
)


plt.tight_layout()
plt.show()
```
